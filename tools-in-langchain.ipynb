```python
# Custom Tools - Method - 1 -> @tool
from langchain_core.tools import tool



# Step 1 - create a function
# Step 2 - add type hint
# Step 3 - add tool decorator

@tool
def multiply(a: int, b: int) -> int:
  """Multiply 2 Integers"""
  return a*b



res = multiply.invoke(
  {
    "a": 5,
    "b": 10
  }
)


print("Result is = ", res)

print('\n')
print("multiply.name -> ", multiply.name)
print("multiply.description -> ", multiply.description)
print("multiply.args -> ", multiply.args)

print('\n')
print("multiply.args_schema.model_json_schema() -> ", multiply.args_schema.model_json_schema())
```

```python
# Custom Tools - Method - 2 -> Structured Tool
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field


class MultiplyInput(BaseModel):
  a: int = Field(required=True, description="The first number to add")
  b: int = Field(required=True, description="The second number to add")


def multiply_func(a: int, b: int) -> int:
  return a*b

multiply_tool = StructuredTool.from_function(
  func=multiply_func,
  name="multiply",
  description="Multiply 2 numbers",
  args_schema=MultiplyInput
)





res = multiply_tool.invoke(
  {
    "a": 5,
    "b": 10
  }
)


print("Result is = ", res)

print('\n')
print("multiply.name -> ", multiply_tool.name)
print("multiply.description -> ", multiply_tool.description)
print("multiply.args -> ", multiply_tool.args)

print('\n')
print("multiply.args_schema.model_json_schema() -> ", multiply_tool.args_schema.model_json_schema())
```

```python
# Custom Tools - Method - 3 -> Base Tool
from langchain_community.tools import BaseTool
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
from typing import Type


class MultiplyInput2(BaseModel):
  a: int = Field(required=True, description="The first number to add")
  b: int = Field(required=True, description="The second number to add")


class MultiplyTool2(BaseTool):
  name: str = "Multiply"
  description: str = "Multiply 2 numbers"
  args_schema: Type[BaseModel] = MultiplyInput2

  def _run(self, a: int, b: int) -> int:
    return a*b




multiply_tool_new = MultiplyTool2()




res = multiply_tool_new.invoke(
  {
    "a": 5,
    "b": 10
  }
)


print("Result is = ", res)

print('\n')
print("multiply.name -> ", multiply_tool_new.name)
print("multiply.description -> ", multiply_tool_new.description)
print("multiply.args -> ", multiply_tool_new.args)

print('\n')
print("multiply.args_schema.model_json_schema() -> ", multiply_tool_new.args_schema.model_json_schema())




```

```python
# Toolkit

from langchain_core.tools import tool

@tool
def add(a:int, b:int) -> int:
  """Add 2 numbers"""
  return a+b

@tool
def multiply(a:int, b:int) -> int:
  """Multiply 2 numbers"""
  return a*b


class MathToolKit:
  def get_tools(self):
    return [add, multiply]


toolkit = MathToolKit()
tools = toolkit.get_tools()


for tool in tools:
  print("-"*10)
  print("tool.name -> ", tool.name)
  print("tool.description -> ", tool.description)
  print("tool.args -> ", tool.args)
  print("-"*10)
```
